```python
from typing import List, TypedDict
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    name:str
    values: List[int]
    operation: str
    result: str
```

```python
def process(state: AgentState)->AgentState:
    """"This function takes values and operation and performs operation"""

    if state['operation']=='+':
        state['result']= f'Hey {state["name"]} the result is {sum(state["values"])}'
    elif state['operation']=='-':
        total=0
        for val in state['values']:
            total-=val
        state['result']=f'Hey {state["name"]} the result is {total}'
    elif state['operation']=='*':
        total=1
        for val in state['values']:
            total*=val
        state['result']=f'Hey {state["name"]} the result is {total}'
    elif state['operation']=='/':
        total=0
        for val in state['values']:
            if val==0:
                state['result']= f"Hry {state["name"]} can't divide by Zero"
                return state
            total/=val
        state['result']=f'Hey {state["name"]} the result is {total}'

    return state
```

```python
graph= StateGraph(AgentState)

graph.add_node("processor",process)
graph.set_entry_point("processor")
graph.set_finish_point("processor")

app=graph.compile()
```

```python
result= app.invoke({'name':'Arman','values':[2,5,6],'operation':'/'})
```
